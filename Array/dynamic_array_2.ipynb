```python
import ctypes
```

```python
class MeraList:
    def __init__(self):
        self.size = 1 # it keeps track of the size of the list
        self.n = 0  # it keeps track of the number of elements in the list
        # create a C type array with size = self.size
        self.A = self.__make_array(self.size)

    def __len__(self):
        return self.n
    
    def append(self, item):
        if self.n == self.size:
            ## Resize
            self.__resize(self.size * 2)
        ## append
        self.A[self.n] = item
        self.n += 1

    def __resize(self, new_capacity):
        # create a new array with new_capacity
        B = self.__make_array(new_capacity)
        self.size = new_capacity
        # copy the content of A into B
        for i in range(self.n):
            B[i] = self.A[i]
        # reassign A
        self.A = B
    def __make_array(self, capacity):
        return (capacity * ctypes.py_object)() ## creates a c type array(fixed, statical) with size capacity
  
```
