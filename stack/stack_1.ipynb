```python
#stack using linked_list
class Node:
  def __init__(self,value):
    self.data = value
    self.next = None
```

```python
class Stack:
  def __init__(self):
    self.Top = None
    self.n = 0

  def is_empty(self):
    return self.Top == None ## returns boolean value if self.Top == None, True else , false

  def __len__(self):
    return self.n

  def push(self, value):
    if self.is_empty == False:
      return "Stack is full"
    new_node = Node(value)

    new_node.next = self.Top
    self.Top = new_node
    self.n = self.n+1
  def peek(self):
    if (self.is_empty()):
      return "stack empty"
    else:
      return self.Top.data
  def pop(self):
    if (self.is_empty()):
      return "stack empty"
    else:
      data = self.Top.data
      self.Top = self.Top.next
      self.n =self.n -1
    return data
  def traverse(self):
    if (self.is_empty()):
      return "stack empty"
    temp = self.Top
    while temp != None:
      print(temp.data)
      temp = temp.next
s = Stack()




```

```python
def reverse_string(text):
  s = Stack()
  for i in text:
    s.push(i)

  res = " "
  while (not s.is_empty()):
    res = res + s.pop()
  print(res)
```
