```python
class Stack:
  def __init__(self, size):
    self.size = size
    self.__Stack = [None] * self.size
    self.Top = -1

  def push(self,value):
    if self.Top == self.size - 1:
      return "OverFlow"
    else:
      self.Top +=1
      self.__Stack[self.Top] = value

  def pop(self):
    if self.Top == -1:
      return "Empty Stack"
    else:
      data = self.__Stack[self.Top]
      # self.Stack[self.Top] = None ---> it can be used but logically after being poped, the stack with size of Top gets considered as a stack but old data remains physically in the stack
      self.Top = self.Top -1
    return data
  def traverse(self):
    if self.Top == -1:
      return "Empty Stack"
    for i in range(self.Top +1):
      print(self.__Stack[i], end = ' ')
  

S = Stack(3)
```
