```python
# Write a program to find the max value in a linked list and replace it with a given value.---replace_max()
# Assume that the linked list is populated with whole numbers and there is only one max value in the linked list--replace_max()
## Write a program for the sum of numbers at odd position.---sum_odd_nodes()
### write a program for reversing a linked list without creating it---reverse()
#### Write a program for removing special symbol from the alphabets or strings
#### Rules
    #1. replace '*' or '/' by a single space
    #2. in case of two consecutive occurences of '*' or '/', replace those two occurence by a single space and  convert teh next character to uppercase
```

```python
class Node:
  def __init__(self,value):
    self.data = value
    self.next = None
```

```python
class LinkedList:
  def __init__(self):
    self.head = None
    self.n = 0

  def __len__(self):
    return self.n

  # def __str__(self):
  #   if self.n == 0:
  #     return "empty linked list(print)"
  #   result = ""
  #   curr = self.head
  #   while curr != None:
  #     result = result + str(curr.data) + "->"
  #     curr = curr.next
  #   return result[:-2]

  def __str__(self):
    if self.n == 0:
      return "empty linked list(print)"
    result = ""
    curr = self.head
    while curr != None:
      result = result + str(curr.data)
      curr = curr.next
    return result

  def append(self,value):
    new_node = Node(value)
    
    if self.head == None:
      self.head = new_node
      self.n = self.n + 1
      return "empty linked list appended"
    curr = self.head
    while curr.next != None:
      curr = curr.next
    curr.next = new_node
    self.n = self.n + 1
    return "value appended"

  def replace_max(self, value):
    curr = self.head
    if self.head == None:
      return "empty"
    max = curr
    
    while curr != None:
      if curr.data > max.data:
        max = curr
      curr = curr.next
    max.data = value

  def sum_odd_nodes(self):
    curr = self.head
    counter = 0
    result = 0
    while curr != None:
      if counter % 2 != 0:
        result = result + curr.data
      counter = counter + 1
      curr = curr.next
    return result

  def reverse(self):
    prev_node = None
    curr_node = self.head
    while curr_node != None:
      next_node = curr_node.next
      curr_node.next = prev_node
      prev_node = curr_node
      curr_node = next_node
    self.head = prev_node

  def change_sent(self):
    temp = self.head
    while temp!= None:

      if temp.data == '*' or temp.data == '/':
        temp.data = ' '
        if temp.next.data == '*' or temp.next.data == '/':
          temp.next.next.data = temp.next.next.data.upper()
          temp.next = temp.next.next
          self.n = self.n - 1
      temp = temp.next

  

L = LinkedList()
word_list = LinkedList()
```
