```python
# creae Node
class Node:
  def __init__(self, value):
    self.value = value
    self.next = None
```

```python
a = Node(1)
b = Node(2)
c = Node(3)
```

```python
id(c)
```

```python
print(a.value, a.next)
```

```python
a.next = b
b.next = c
```

```python
print(int(id(a.next)), int(id(b.next)), int(id(c.next)))
```
